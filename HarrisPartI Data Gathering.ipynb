```python
# Since it's my first time coding, Michelle recommended to check if we had these things installed and she created a nice little code that says wether it is installed or not!
import importlib.util
import sys
import subprocess

def check_and_install(package_name):
    # Check if the package is installed
    spec = importlib.util.find_spec(package_name)
    
    if spec is None:
        # Install if it is not found
        print(f" '{package_name}' not found. Installing now...")
        subprocess.check_call([sys.executable, "-m", "pip", "install", package_name])
    else:
        print(f" '{package_name}' is already installed.")

check_and_install("numpy")
check_and_install("matplotlib")
check_and_install("cmasher")
check_and_install("astropy")
check_and_install("scipy")

check_and_install("pandas") # I need this
```

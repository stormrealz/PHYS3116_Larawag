```python
import numpy as np
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, figsize=(6, 5))
FeHvsAge = axes.scatter(filtered_FeH, filtered_Age, c=filtered_R_G,
                        marker='o', s=30, cmap='viridis')
axes.set_xlabel('$FeH$ ($Dex$)')
axes.set_ylabel('$Age$ ($Gyr$)')
fig.colorbar(FeHvsAge, ax=axes, label='R_gc')

# Create a exponential curve to divide the accreted and in-situ GC's
def model(x, a, b, c):
    return c - a * np.exp(b * x)

a, b, c = 11.6, 1.22, 14.6

x_fit = np.linspace(-2.4, -0.48, 300)
axes.plot(x_fit, model(x_fit, a, b, c), color='red', lw=2,
          label=f'$Age = {c:.1f} - {a:.1f}\\,e^{{{b:.1f}[Fe/H]}}$') # exponential function and a function of FeH
axes.legend(loc='lower left', fontsize=8)
plt.show()
```

```python
x = np.asarray(filtered_FeH, dtype=float)
y = np.asarray(filtered_Age, dtype=float)
rgc = np.asarray(filtered_R_G, dtype=float)

redcutoff = model(x, a, b, c)        # age the red line gives at each point's [Fe/H]
in_range = (x >= -2.4) & (x <= -0.48)  # so it doesn't go outside the dataset
below = (y < redcutoff) & in_range # things below the red line and between -2.4 and -0.48

below_df = pd.DataFrame({ 
    'FeH': x[below], 
    'Age': y[below],
    'R_gc': rgc[below],
    'Line_Age': redcutoff[below],
    'Offset': y[below] - redcutoff[below],   # negative = below the line
}).sort_values('Offset').reset_index(drop=True)

print(f"{below.sum()} of {len(x)} points are below the line")
print(below_df)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 0. Convert to pandas and see what the columns are called
df = filtered_catalogue_will.to_pandas()
df.columns = [str(c).strip() for c in df.columns]
print(df.columns.tolist())

# 1. Rebuild the Cluster label from the first two (NGC) columns
def clean(s):
    return (s.astype('string').str.strip()
             .str.replace(r'\.0$', '', regex=True)
             .replace({'': pd.NA, 'nan': pd.NA, '--': pd.NA}))

ngc = clean(df.iloc[:, 0]).fillna(clean(df.iloc[:, 1]))
df['Cluster'] = ngc.where(~ngc.str.isdigit().fillna(False), 'NGC ' + ngc)

# 2. Mask: True for the clusters below the red line
below_names = below['Cluster'].tolist()
mask = df['Cluster'].isin(below_names).to_numpy()
print(f"{mask.sum()} below the line, {(~mask).sum()} others")   # expect 36 below

# 3. Table of the below-line clusters with their values
below_table = df[mask].copy()
below_table['Offset'] = below_table['Cluster'].map(below.set_index('Cluster')['Offset'])
print(below_table.sort_values('Offset')[['Cluster', 'FeH', 'Age', 'R_G', 'X', 'Z', 'Offset']]
      .to_string(index=False))

# 4. X-Z plot: red = below the line, blue = everything else
x = np.asarray(filtered_FeH, dtype=float)
z = np.asarray(filtered_Age, dtype=float)
assert len(x) == len(mask), f"{len(x)} plotted points vs {len(mask)} table rows"


fig, axes = plt.subplots(1, figsize=(6, 6))

# Age = c - a*exp(b*[Fe/H]), tuned by eye to match the drawn curve
def model(x, a, b, c):
    return c - a * np.exp(b * x)

a, b, c = 11.6, 1.22, 14.6

x_fit = np.linspace(-2.4, -0.1, 300)
axes.plot(x_fit, model(x_fit, a, b, c), color='red', lw=2,
          label=f'$Age = {c:.1f} - {a:.1f}\\,e^{{{b:.1f}[Fe/H]}}$')

axes.scatter(x[~mask], z[~mask], color='blue', s=40, label='Other')
axes.scatter(x[mask], z[mask], color='red', s=40, label='Below the line')
axes.set_xlabel('$FeH$, $Dex$') # x-axis label
axes.set_ylabel('$Age$, $Gyr$') # y-axis label
axes.set_ylim(7.5, 15)

axes.legend(fontsize=8)

plt.show()
```
